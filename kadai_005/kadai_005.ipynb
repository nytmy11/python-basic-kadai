# 変数を使って四則演算しよう

台形の面積を求めて、単位（cm²）をつけて出力する。

- 上辺：10cm
- 下辺：20cm
- 高さ：5cm
- 公式：台形の面積 =（上辺 + 下辺）× 高さ ÷ 2


```python
# 台形の面積を求める
upper_base = 10  # 上辺(cm)
lower_base = 20  # 下辺(cm)
height = 5       # 高さ(cm)

# 台形の面積 =（上辺 + 下辺）× 高さ ÷ 2
trapezoid_area = (upper_base + lower_base) * height / 2

print(str(trapezoid_area) + "cm²")
```
