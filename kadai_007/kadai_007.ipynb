# リスト/ディクショナリを使いこなそう

曜日ごとの天気予報をリスト `array` とディクショナリ `dictionary` で作成し、
**水曜日**の天気予報をそれぞれから取り出して出力する。

- リストは順序（インデックス 0〜6）で管理する → 水曜日は `array[2]`
- ディクショナリはキー（mon〜sun）で管理する → 水曜日は `dictionary["wed"]`


```python
# 曜日ごとの天気予報をリストで作成する
array = [
    "月曜日は晴れです",
    "火曜日は雨です",
    "水曜日は晴れです",
    "木曜日は晴れです",
    "金曜日は曇りです",
    "土曜日は曇りのち雨です",
    "日曜日は雷雨です",
]

# 曜日ごとの天気予報をディクショナリで作成する
dictionary = {
    "mon": "晴れ",
    "tue": "雨",
    "wed": "晴れ",
    "thu": "晴れ",
    "fri": "曇り",
    "sat": "曇りのち雨",
    "sun": "雷雨",
}

# 水曜日の天気予報を取り出して出力する
print(array[2])       # リストはインデックス（0始まり）で取り出す
print(dictionary["wed"])  # ディクショナリはキーで取り出す
```
