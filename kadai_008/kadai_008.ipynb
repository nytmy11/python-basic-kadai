# 条件分岐のif文で処理を切り替えよう

8章「条件分岐のif文を理解しよう」の課題です。変数 `var` の値を if 文で判定して、FizzBuzz 問題の結果を出力します。

| 条件 | 出力 |
|---|---|
| 3の倍数かつ5の倍数 | `FizzBuzz` |
| 3の倍数 | `Fizz` |
| 5の倍数 | `Buzz` |
| どれにも該当しない | `var` の値 |

**判定の順番がポイント**です。`if var % 3 == 0` を先に書いてしまうと、15 のように両方の倍数である値が最初の分岐に入ってしまい `Fizz` としか出力されません。そのため、条件が最も厳しい「3の倍数かつ5の倍数」を先頭に置き、`and` で2つの条件を組み合わせて判定しています。

```python
# 判定対象の数値を変数varに代入する（varは正の整数）
var = 15

# 3の倍数かつ5の倍数の場合を最初に判定する
# （先に「3の倍数」を判定すると、15のときにFizzだけが出力されてしまうため）
if var % 3 == 0 and var % 5 == 0:
    print("FizzBuzz")
elif var % 3 == 0:
    print("Fizz")
elif var % 5 == 0:
    print("Buzz")
else:
    print(var)
```
