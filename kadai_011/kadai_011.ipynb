# 繰り返し処理を使いこなそう

11章までに学んだ知識を使って、リスト `array` の中身を先頭から全て取り出して出力します。同じ結果を **for文** と **while文** の2通りで書きました。

## forとwhileの違い

| | for | while |
|---|---|---|
| 繰り返しの条件 | リストの要素を順に取り出し、なくなったら終了 | 条件式が `True` の間ずっと繰り返す |
| カウンタ変数 | 不要（Pythonが要素を渡してくれる） | 自分で用意して自分で進める（`i += 1`） |
| 向いている場面 | 繰り返す回数・対象があらかじめ決まっているとき | 終了条件が実行してみないと分からないとき |

今回のように「リストの全要素を処理する」ケースは回数が決まっているので **for文のほうが素直**です。while文では `i` の初期化と `i += 1` を書き忘れると無限ループになるため、`i < len(array)` で範囲を超えないようにしています。

```python
array = ["水", "金", "地", "火", "木", "土", "天", "海", "冥"]

# for文：リストから要素を1つずつ直接取り出す
for planet in array:
    print(planet)

# while文：インデックスを自分で進めながら取り出す
i = 0
while i < len(array):
    print(array[i])
    i += 1
```
