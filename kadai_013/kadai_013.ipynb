# 関数を使いこなそう

12〜13章で学んだ知識を使って、商品購入時の消費税を計算する関数を作ります。

## 関数の仕様

| | 内容 |
|---|---|
| 関数名 | `add_tax` |
| 第1引数 | `price` — 商品の金額（今回は `110`） |
| 第2引数 | `tax_rate` — 消費税率（%、今回は `10`） |
| 戻り値 | 消費税を加えた金額 |

`return` で計算結果を返しているので、呼び出し側で `result` に受け取ってから `print` で出力できます。関数の中で `print` してしまうと値を使い回せないため、「結果を返す関数」の指示どおり戻り値を使いました。

## 計算式について

税込金額は `price * (1 + tax_rate / 100)` とも書けますが、この式だと `110 * 1.1` の浮動小数点計算で `121.00000000000001` という誤差が出ます。そこで **消費税額を先に求めてから本体価格に足す**（`price + price * tax_rate / 100`）形にしました。`110 * 10 / 100` は誤差なく `11.0` になるため、結果も `121.0` ときれいに求まります。課題の指示どおり、切り捨てや四捨五入はしていません。

```python
# 消費税を加えた金額を返す関数
# 第1引数 price    : 商品の金額
# 第2引数 tax_rate : 消費税率（%）
def add_tax(price, tax_rate):
    tax = price * tax_rate / 100  # 消費税額を求める
    return price + tax            # 税込金額を戻り値として返す


# 商品の金額110円、消費税10%で関数を呼び出す
result = add_tax(110, 10)

print(result)
```
