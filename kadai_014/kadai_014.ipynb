# 変数スコープを正しく使おう

14章で学んだ変数スコープの知識を使って、エラーが発生するプログラムを修正します。

## エラーの原因

元のコードは次のようになっていました。

```python
price1 = 100
price2 = 200

def total():
    tax = 1.1
    return price1 + price2

print (total() * tax)
```

これを実行すると、最終行で次のエラーになります。

```
NameError: name 'tax' is not defined
```

`tax = 1.1` は **関数 `total()` の中で定義されたローカル変数**です。ローカル変数が生きているのはその関数の中だけで、関数を抜けると消えてしまいます。そのため関数の外にある `print(total() * tax)` からは `tax` が見えず、「そんな名前は定義されていない」と怒られます。

一方で `price1` と `price2` は関数の外（グローバルスコープ）で定義されているため、関数 `total()` の中からでも参照できます。**外から中は見えるが、中から外は見えない**——これがグローバルスコープとローカルスコープの関係です。

| 変数 | 定義した場所 | スコープ | 関数の外から参照 |
|---|---|---|---|
| `price1` / `price2` | 関数の外 | グローバル | できる |
| `tax`（修正前） | 関数 `total()` の中 | ローカル | **できない** → NameError |

## 修正した点

`tax = 1.1` を **関数の外（グローバルスコープ）に移動**しました。これで `print()` からも `tax` を参照できるようになり、`(100 + 200) * 1.1 = 330.0` が正しく出力されます。

なお、`tax` を関数の中に置いたまま `return (price1 + price2) * tax` として税込金額ごと返す書き方でも解決できます。こちらは税率が関数の外に漏れないぶん設計としてはきれいですが、今回はスコープの違いをはっきり示すため、変数を外に出す方法で修正しました。

````python
price1 = 100
price2 = 200
tax = 1.1  # グローバルスコープに移動する（関数の外でも参照できるようにするため）


def total():
    # price1 と price2 はグローバル変数なので、関数の中から参照できる
    return price1 + price2


# tax がグローバルスコープにあるので、関数の外でも使える
print(total() * tax)
````
