# クラスを使いこなそう（Pythonの基礎を学ぼう・15章の課題）

名前(`name`)と年齢(`age`)の属性を持つ `Human` クラスを定義し、
その2つを標準出力するメソッド `printinfo` を追加した。

## クラスの構成

| 要素 | 名前 | 役割 |
| --- | --- | --- |
| コンストラクタ | `__init__(self, name, age)` | インスタンス生成時に呼ばれ、属性に初期値をセットする |
| 属性 | `self.name` | 名前を保持する |
| 属性 | `self.age` | 年齢を保持する |
| メソッド | `printinfo(self)` | 自身の `name` と `age` を `print` で出力する |

- **属性**はインスタンスが持つ「データ」、**メソッド**はインスタンスが持つ「操作」。
- `self` は「そのインスタンス自身」を指す。`self.name` と書くことで、
  インスタンスごとに別々の値を保持できる（`human1` の名前と `human2` の名前は混ざらない）。
- 生成したインスタンスは変数 `human1` / `human2` に代入して使用している。

## 実行結果

`Human("山田太郎", 25)` と `Human("鈴木花子", 32)` の2つのインスタンスを作り、
それぞれで `printinfo()` を呼び出すと、各インスタンス自身の属性が出力される。

```python
# 人を表す Human クラス
class Human:
    # コンストラクタ：インスタンス生成時に name と age の属性をセットする
    def __init__(self, name, age):
        self.name = name  # 属性：名前
        self.age = age    # 属性：年齢

    # メソッド：自身の名前と年齢を標準出力する
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# インスタンスを変数に代入して使用する
human1 = Human("山田太郎", 25)
human1.printinfo()

# 別のインスタンスを作れば、それぞれが自分の属性を持つ
human2 = Human("鈴木花子", 32)
human2.printinfo()
```
