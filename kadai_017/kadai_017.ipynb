名前(`name`)と年齢(`age`)の属性を持つ `Human` クラスを作り、
年齢が20以上かどうかで出力を切り替える `check_adult` メソッドを追加した。
インスタンスを複数生成してリストに入れ、リストの要素数分だけ `check_adult` を呼び出している。

## 使用した学習項目

合格基準で指定されている6項目を、それぞれ次の形で使っている。

| 学習項目 | 使っている箇所 |
| --- | --- |
| リスト | `profiles`（もとデータ）と `humans`（インスタンスの入れ物） |
| ディクショナリ | `{"name": "山田太郎", "age": 25}` — 名前と年齢をキーで持つ |
| 条件分岐 | `check_adult` 内の `if self.age >= ADULT_AGE: ... else: ...` |
| 繰り返し処理 | `create_humans` 内の `for profile in profiles`、最後の `for human in humans` |
| 関数 | `create_humans(profiles)` — 引数を受け取り、リストを戻り値として返す |
| 変数のスコープ | `ADULT_AGE` をグローバルスコープに置き、メソッドの中から参照している |
| クラス | `Human`（属性 `name` / `age`、メソッド `check_adult`） |

## 設計のポイント

- **判定の境界値をグローバル定数にした。**
  `20` をメソッドの中に直接書くこともできるが、`ADULT_AGE` という名前でグローバルスコープに
  置いておけば、成人年齢が変わったときに1か所だけ直せばよい。
  グローバル変数は関数やメソッドの**中から参照できる**という性質を利用している。
- **もとデータと処理を分けた。**
  名前と年齢はディクショナリのリストとして持ち、そこから `Human` インスタンスを作る処理は
  `create_humans` 関数に切り出した。データが増えても `profiles` に1行足すだけで済む。
- **境界値も確認できるようにした。**
  ちょうど20歳の「佐藤次郎」を入れてある。判定は `>=` なので20歳は「大人です」になる。

## 実行結果

4人分のインスタンスについて、それぞれの年齢に応じた判定が出力される。

```python
# 成人とみなす年齢
# グローバルスコープに置くことで、クラスのメソッドの中からも参照できる
ADULT_AGE = 20


class Human:
    """名前と年齢を持つ人を表すクラス"""

    # コンストラクタ：インスタンス生成時に属性へ初期値をセットする
    def __init__(self, name, age):
        self.name = name  # 属性：名前
        self.age = age    # 属性：年齢

    # メソッド：成人かどうかを判定して標準出力する
    def check_adult(self):
        # 条件分岐：グローバル変数 ADULT_AGE と比較する
        if self.age >= ADULT_AGE:
            print(f"{self.name}さん（{self.age}歳）は大人です")
        else:
            print(f"{self.name}さん（{self.age}歳）は大人ではありません")


# 関数：ディクショナリのリストから Human インスタンスのリストを作って返す
def create_humans(profiles):
    humans = []                       # 空のリストを用意する
    for profile in profiles:          # 繰り返し処理でディクショナリを1件ずつ取り出す
        humans.append(Human(profile["name"], profile["age"]))
    return humans                     # 戻り値としてリストを返す


# もとデータ：名前と年齢をディクショナリで持ち、それをリストにまとめる
profiles = [
    {"name": "山田太郎", "age": 25},
    {"name": "鈴木花子", "age": 17},
    {"name": "佐藤次郎", "age": 20},
    {"name": "田中三郎", "age": 8},
]

# 関数を呼び出して、インスタンスのリストを受け取る
humans = create_humans(profiles)

# リストの要素数分だけ check_adult メソッドを呼び出す
for human in humans:
    human.check_adult()
```
